```python
import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split

os.makedirs("../datasets/clean", exist_ok=True)
```

```python
train_path = "../datasets/train.csv"
test_path = "../datasets/test.csv"  

# Load datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print(f"Train shape: {train_df.shape}")
print(f"Test shape: {test_df.shape}")
```

```python
# Drop unnormalized rows
train_df.dropna(subset=["text"], inplace=True)
train_df = train_df[train_df["text"].str.strip().astype(bool)]

# Reset index
train_df.reset_index(drop=True, inplace=True)

# Rename target to label
train_df.rename(columns={"target": "label"}, inplace=True)

# Lowercase text
train_df["text"] = train_df["text"].str.strip()

# Sanity check
print(train_df["label"].value_counts())
print(train_df.head())
```

```python
# Clean text column + keep original index
test_df.dropna(subset=["text"], inplace=True)
test_df = test_df[test_df["text"].str.strip().astype(bool)]
test_df.reset_index(drop=True, inplace=True)
test_df["text"] = test_df["text"].str.strip()

print(test_df.head())
```

```python
# Split Train/Test
train_split, temp_split = train_test_split(
    train_df, test_size=0.10, stratify=train_df["label"], random_state=42
)

# Split Val/Test from the 10% temp
val_split, test_split = train_test_split(
    temp_split, test_size=0.5, stratify=temp_split["label"], random_state=42
)

print(f"Final Split Sizes:")
print(f"Train: {len(train_split)}")
print(f"Val: {len(val_split)}")
print(f"Test: {len(test_split)}")
```

```python
# Save cleaned datasets
train_split.to_csv("../datasets/clean/train.csv", index=False)
val_split.to_csv("../datasets/clean/val.csv", index=False)
test_split.to_csv("../datasets/clean/test.csv", index=False)
test_df.to_csv("../datasets/clean/unlabeled_test.csv", index=False)

print("Saved all cleaned data to /datasets/clean/")
```
